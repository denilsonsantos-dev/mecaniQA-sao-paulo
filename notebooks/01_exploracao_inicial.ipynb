# Indexação Temporal e Inspeção Inicial

O que a gente quer fazer: importar o dataset.csv e dar uma olhada nele (ver as colunas, quantos registros tem, se falta dado, etc). Esse notebook mostra o código que a gente usou no primeiro encontro sobre como organizar dados de tempo no Pandas. Pra saber mais, olha o README.md.

```python
import pandas as pd
from pathlib import Path

pd.set_option('display.max_columns', None)
```

(1): Importando o dataset. Pra o Pandas entender que é um dado de tempo, a gente precisa fazer duas coisas: mudar a coluna de data pra datetime64 e fazer dessa coluna o índice do DataFrame. A coluna de data no nosso dataset se chama Data.

```python
DATA_PATH = Path('..') / 'data' / 'raw' / 'dataset.csv'
COLUNA_DATA = 'Data'

df = pd.read_csv(
    DATA_PATH,
    parse_dates=[COLUNA_DATA],
    index_col=COLUNA_DATA
)

df = df.sort_index()

df.head()
```

```python
type(df.index)
```
